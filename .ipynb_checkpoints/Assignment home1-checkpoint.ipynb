QUESTION 1. 

Write a program that takes two numbers (a = 15 , b = 4) and prints their:

Sum

Difference

Product

Division

Floor Division

Modulus

```python
a = 15
b = 4

print("Sum:", a + b)
print("Difference:", a - b)
print("Product:", a * b)
print("Division:", a / b)
print("Floor Division:", a // b)
print("Modulus:", a % b)
```

QUESTION 2.

```python
num1 = 10
num2 = 5.6
num3 = 4.43
num4 = '23'

print("Type of num1:", type(num1))
print("Type of num2:", type(num2))
print("Type of num3:", type(num3))
print("Type of num4:", type(num4))

print("Integer to float:", float(num1))
print("Float to integer:", int(num2))
print("Float to string:", str(num3))
print("String to integer:", int(num4))
```

QUESTION 10.

Print all two results

QUESTION 11. 

You have a product code that follows the format "SKU-12345-XYZ".

Find the first position of the dash (-).

Count the number of dashes (-) in the product code.

Remove the "SKU-" prefix and "-XYZ" suffix from the product code, Extract the numeric part 12345

product_code = "SKU-12345-XYZ"

Hint: Use String Methods

```python
product_code = "SKU-12345-XYZ"

print("First position of dash is:", product_code.find("-"))

print("Total number of dashes is:", product_code.count("-"))

numeric_part = product_code.replace("SKU-", "").replace("-XYZ", "")

print("Numeric part of the Product is:", numeric_part)
```

QUESTION 14. 

text = " Data Analysis with Python is very powerful "

Perform the following tasks:

- Remove extra spaces using strip().
- Convert the string into title case using title().
- Replace "Powerful" with "Important".
- Find the position of the word "Python".

```python
text = "    Data Analysis with Python is very powerful    "

text = text.strip()
print("Removed Spaces:", text)

text = text.title()
print("Title Case:", text)

text = text.replace("Powerful", "Important")
print("Final text is :", text)

position = text.find("Python")
print("Position of Python is:", position)
```

QUESTION 15.
    
Given the list: data = [12, 45, 2, 67, 34, 89, 23, 90]

Perform the following tasks:

- Sort the list in ascending order.
- Remove the highest value from the list.
- Insert the value 50 at index 3.". 
- Finally, reverse the updated list and display the result.

```python
data = [12, 45, 2, 67, 34, 89, 23, 90]

data.sort()
print("Sorted data:", data)

data.remove(max(data))
print("Remove highest value:", data)

data.insert(3, 50)
print("Inserted data:", data)

data.reverse()
print("Reversed data", data)
```
